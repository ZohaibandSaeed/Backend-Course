```python
from langchain_core.tools import tool
```

```python
@tool
def mulitply(first_number: int, second_number: int) -> int:
  """multiply two integer numbers"""
  return first_number*second_number

@tool
def addition(first_number: int, second_number: int) -> int:
  """adding two integer numbers"""
  return first_number+second_number
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(
    model="gemini-3.1-flash-lite",
    api_key="AQ.abc"
)

response = llm.invoke("what is the distance between moon and earth tell me in one line")

print(response)
```

```python
data = response.tool_calls
data_list = data[0]

args = data_list["args"]
name = data_list["name"]
```

```python

def get_tool_description(tool_name):
    # Aap ka data variable
    data = llm_with_tool.kwargs.get('tools', [])

    # List mein check karne kay liye loop
    for tool in data:
        if tool['function']['name'] == tool_name:
            return tool['function']['description']

    return "Tool nahi mila!"  # Agar naam ghalat ho

description = get_tool_description(name)

print(description)
```
